# Análise de Acidentes nas Rodovias da Bahia - PRF

Este notebook explora os dados abertos da [Polícia Rodoviária Federal (PRF)](https://www.gov.br/prf/pt-br/acesso-a-informacao/dados-abertos/dados-abertos-da-prf), referentes aos registros de acidentes agrupados por pessoa, de 2019 a 2025.

O objetivo é investigar os acidentes ocorridos nas rodovias do estado da Bahia, utilizando os dados nacionais para selecionar os registros correspondentes ao estado.

## Importação das bibliotecas utilizadas no Projeto:

```python
import pandas as pd
```

Os arquivos `.csv` de dados de acidentes da PRF estão estruturados com valores separados por **ponto e vírgula** `;`, o que justifica o parâmetro `sep=';'`

Esses arquivos utilizam o formato numérico **brasileiro**. Significa que os números decimais são separados por **vírgula** `,` ao invés de **ponto** `.`. O que justifica o uso do parâmetro `decimal=','`

Além disso, há acentuação e caracteres especiais no idioma brasileiro, justificando o uso de `encoding='latin-1'` (ou _encoding='cp1252'_). Já que a utilização do formato _UTF-8_, apresentou o erro `byte 0xe1`.

## Exemplo de carregamento de dados (csv 2025):

```python
# Lendo o csv de 2025
df = pd.read_csv("../data/raw/acidentes2025.csv", sep=';', decimal=',', encoding='latin-1')

# df = pd.read_csv("../data/raw/acidentes2025.csv", sep=';', decimal=',', encoding='utf-8')
# Caso o comentário da linha acima seja removido, o código apresentará erro devido à utilização do 'utf-8'

df.head()
```

## Contagem de Linhas e Colunas:

```python
qtd = []
total_lin = 0  # Variável que conta o total de linhas

# Loop de leitura dos CSVs
for ano in range(2019, 2026):
    path = f"../data/raw/acidentes{ano}.csv"  # Automatizando a mudança de arquivo CSV
    df_ano = pd.read_csv(path, sep=';', decimal=',', encoding='latin-1')
    
    # Adicionando as informações (ano, linhas e colunas) à lista
    qtd.append({
        "Ano": ano,
        "Linhas": df_ano.shape[0],
        "Colunas": df_ano.shape[1]
    })

# Transformando as listas em um DataFrame e definindo o index
escopo = pd.DataFrame(qtd).set_index("Ano")
print(f"{escopo['Linhas'].sum()} Linhas Totais")
escopo
```

## Carregando a base de dados completa:

Concatenando todos os dados (2019 a 2025) em um único DataFrame e adicionando a coluna **ano** a cada um dos conjuntos de dados, para facilitar a identificação.

Agora, o novo DataFrame (`df`) deve conter **1249899** linhas × **36 colunas**

```python
# Lista que armazenará o conjunto de DataFrames
lista = []

# Loop de leitura dos CSVs
for ano in range(2019, 2026):
    path = f"../data/raw/acidentes{ano}.csv"  # Automatizando a mudança de arquivo CSV

    df_ano = pd.read_csv(path, sep=';', decimal=',', encoding='latin-1')
    df_ano['ano'] = ano  # Adicionando a coluna 'ano'

    lista.append(df_ano)

df = pd.concat(lista, ignore_index=True)  # Transformando a lista de DataFrames em um DataFrame único
df
```

## Padronização do tipo da coluna `id`

A coluna de ID do DataFrame geral foi lida com o formato de número **decimal**, quando deveriam ser números **inteiros**, de acordo com o dicionário de dados da PRF.

Ao analisar todos os arquivos de acidentes, é possível notar que apenas os CSVs dos anos de 2022, 2023 e 2025 possuem a coluna 'id' com o formato `int64`, sendo que as demais apresentam o formato `float64`.

```python
print("ANO | COLUNA 'id'\n")

# Loop para automatizar a leitura dos arquivos
for ano in range(2019, 2026):
    caminho = f"../data/raw/acidentes{ano}.csv"
    df_ano = pd.read_csv(caminho, sep=';', decimal=',', encoding='latin-1')
    print(ano, df_ano['id'].dtype)  # Retorna o TIPO da coluna 'id' em seu respectivo 'ano'
```

## Identificando se não há nulos na coluna `id`:

```python
# Retornando a soma de valores nulos na coluna de ID
print(df["id"].isna().sum())
```

## Confirmando que todos os valores de ID são originalmente Inteiros:

```python
# Se todos os restos de divisão por 1 forem 0, então podemos ignorar as casas decimais de todos os valores
print((df["id"] % 1 == 0).all())
```

## Convertendo a coluna 'id' de `float64` (decimal) para `int64` (inteiro):
Como a célula acima confirma que nenhum dos IDs são decimais, a fim de gerar uma padronização, a coluna 'id' será convertida para valores inteiros.

```python
df['id'] = df['id'].astype("int64")
df['id'].dtype
```

## Quantas Linhas e quantos Acidentes por ano?
PS: _**Linhas** = linhas por ano_ | _**Acidentes** = IDs por ano_

```python
# Agrupando as linhas por 'ano' e aplicando função de agregação
df.groupby('ano').agg(
    Linhas = ('id', 'size'),  # Coluna 'Linhas': Seleciona a coluna 'id' e conta as linhas totais
    Acidentes = ('id', 'nunique')  # Coluna 'Acidentes': Seleciona a coluna 'id' e conta os valores distintos para cada ano
)
```

De acordo com o resultado acima, em todos os anos a quantidade de linhas totais é cerca de **2,6 a 2,7 vezes** a quantidade de `Acidentes`. Ex: _(2019) 176104 / 67558 ≈ 2,6067_

Portanto, uma linha representa uma **pessoa**, e não um acidente. Logo, o número de acidentes é calculado por `COUNT (DISTINCT id)`, por exemplo, nunca pelo total de linhas em um ano.

## Quais colunas possuem Nulos e quantos são, por Ano?

```python
# Soma dos Nulos por tipo, agrupado por Ano e com os eixos invertidos (colunas viraram linhas)
nulos = df.isna().groupby(df['ano']).sum().T
# Exclui as linhas onde não há nenhum nulo
nulos[nulos.sum(axis=1) > 0]
```

Essa tabela demonstra que há valores **nulos** dentre as informações de: `tipo_acidente`, `classificacao_acidente`, `tipo_veiculo`, `tipo_envolvido`, `estado_fisico`, `sexo`, `regional`, `delegacia` e `uop`.

Inclusive, é possível notar que os valores nulos se repetem, a cada ano, nas colunas de `tipo_envolvido`, `estado_fisico` e `sexo`, o que será analisado mais adiante.

## Os nulos de `tipo_envolvido`, `estado_fisico` e `sexo` estão nas mesmas linhas?

É necessário conferir isso, já que, apesar de apresentarem as mesmas quantidades em todos os anos, a distribuição desses valores dentre as colunas pode ser diferente.

```python
# Seleciona apenas as colunas a serem comparadas
col_nulos = df[['tipo_envolvido', 'estado_fisico', 'sexo']]

# Conferindo as linhas onde os valores das 3 colunas são nulos
val_1 = col_nulos.isna().all(axis=1).sum()
# Conferindo as linhas onde, pelo menos, 1 dos valores das 3 colunas é nulo
val_2 = col_nulos.isna().any(axis=1).sum()

print('Os valores são iguais?', val_1 == val_2)
print(val_1, val_2)
```

```python
# Utilizando a tabela 'nulos', criada acima, para comparar a quantidade de valores nulos
print(nulos.loc['tipo_envolvido'].sum())
```

Utilizando `.all(axis=1).sum()` e `.any(axis=1).sum()`, obtive **109.623** em ambos os casos.

A lógica foi comparar a quantidade de linhas em que as três colunas são nulas simultaneamente (`all`) com a quantidade de linhas em que pelo menos uma delas é nula (`any`). Após isso, também comparei com o total de valores nulos na tabela "nulos".

Como os resultados foram iguais, concluí que os valores nulos de `tipo_envolvido`, `estado_fisico` e `sexo` estão exatamente nas mesmas linhas.
